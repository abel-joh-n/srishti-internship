```python
import numpy as np
# Basic Math
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

```python
print(a + b)
print(a * b)
print(np.sin(a))
```

```python
# Matrix Multiplication
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
```

```python
print(np.dot(A, B))
print(A @ B)
```

```python
# Aggregation
arr = np.array([1, 2, 3, 4])
```

```python
print(arr.sum())
print(arr.mean())
print(arr.max())
print(arr.min())
print(arr.std())
```
