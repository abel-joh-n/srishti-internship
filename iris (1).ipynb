```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["Species"] = iris.target

df.head()
```

```python
print("Dataset shape:", df.shape)

df.info()
```

```python
# Statistical summary
df.describe()
```

```python
# Check for missing values
print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
# Dataset Duplicate Value Count
dup = df.duplicated().sum()
print(f'number of duplicated rows are {dup}')
```

```python
df.Species.value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Species'] = le.fit_transform(df['Species'])
unique_species = df['Species'].unique()
print("Encoded Species Values:")
print(unique_species)
```

```python
df
```

```python
# Species distribution
print("\nSpecies distribution:")
print(df['Species'].value_counts())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

sns.histplot(df['sepal length (cm)'], kde=True, ax=axes[0, 0], color='green')
axes[0, 0].set_title('Sepal Length Distribution')

sns.histplot(df['sepal width (cm)'], kde=True, ax=axes[0, 1], color='skyblue')
axes[0, 1].set_title('Sepal Width Distribution')

sns.histplot(df['petal length (cm)'], kde=True, ax=axes[1, 0], color='orange')
axes[1, 0].set_title('Petal Length Distribution')

sns.histplot(df['petal width (cm)'], kde=True, ax=axes[1, 1], color='red')
axes[1, 1].set_title('Petal Width Distribution')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df.drop('Species', axis=1))
plt.title('Boxplot of All Numerical Features')
plt.ylabel('Centimeters (cm)')
plt.show()
```

```python
# Create the figure
plt.figure(figsize=(8, 6))

species = df['Species'].unique()
colors = ['red', 'green', 'blue']

# Scatter plot for each species
for i in range(len(species)):
    x = df[df['Species'] == species[i]]

    plt.scatter(
        x['sepal length (cm)'],
        x['sepal width (cm)'],
        c=colors[i],
        label=species[i]
    )

# Add labels and legend
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Sepal Length vs Width by Species')
plt.legend()

# Show plot
plt.show()
```

```python
# Calculate correlation
corr = df.drop('Species', axis=1).corr()

# Plot heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    center=0,
    fmt=".2f"
)

plt.title('Feature Correlation Heatmap')
plt.show()
```

```python
# Create subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Plot distributions by species
sns.boxplot(x='Species', y='sepal length (cm)', data=df, ax=axes[0, 0], palette='Set2')
axes[0, 0].set_title('Sepal Length by Species')

sns.boxplot(x='Species', y='sepal width (cm)', data=df, ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title('Sepal Width by Species')

sns.boxplot(x='Species', y='petal length (cm)', data=df, ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Petal Length by Species')

sns.boxplot(x='Species', y='petal width (cm)', data=df, ax=axes[1, 1], palette='Set2')
axes[1, 1].set_title('Petal Width by Species')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.violinplot(
    x='Species',
    y='petal length (cm)',
    data=df,
    inner='quartile',
    palette='pastel'
)

plt.title('Petal Length Distribution by Species')
plt.show()
```
